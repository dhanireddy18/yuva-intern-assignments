## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

## 2. Load the Dataset

```python
DATA_PATH = Path('../data/student_performance.csv')
df = pd.read_csv(DATA_PATH)
df
```

## 3. Inspect and Clean the Data

```python
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
df.describe()
```

## 4. Descriptive Statistics

```python
marks = df['Marks']
print('Mean:', marks.mean())
print('Median:', marks.median())
print('Mode:', marks.mode().tolist())
print('Variance:', marks.var(ddof=0))
print('Standard deviation:', marks.std(ddof=0))
print('Percentiles:', np.percentile(marks, [25, 50, 75]))
```

## 5. Correlation

```python
correlation = df['Study_Hours'].corr(df['Marks'])
print('Study Hours vs Marks correlation:', correlation)
```

## 6. Performance Classification and Filtering

```python
def classify_marks(mark):
    if mark >= 75:
        return 'High'
    elif mark >= 60:
        return 'Medium'
    return 'Low'

df['Performance_Level'] = df['Marks'].apply(classify_marks)
display(df.sort_values('Marks', ascending=False))
print('\nHigh performers:')
display(df[df['Marks'] >= 75])
```

## 7. Grouping

```python
df.groupby('Performance_Level')[['Marks','Attendance','Study_Hours']].mean()
```

## 8. Visualizations

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Study_Hours'], df['Marks'])
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Marks'], bins=5)
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.title('Distribution of Marks')
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.bar(df['Student'], df['Marks'])
plt.xlabel('Student')
plt.ylabel('Marks')
plt.title('Student Marks')
plt.show()
```

## 9. Save Processed Data

```python
output_path = Path('../outputs/processed_student_performance.csv')
df.to_csv(output_path, index=False)
print('Saved:', output_path)
```
